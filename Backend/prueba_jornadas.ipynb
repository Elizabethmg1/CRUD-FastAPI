```python
import requests

BASE_URL = "http://localhost:8000"
```

## Listar todas las jornadas

```python
r = requests.get(f"{BASE_URL}/jornadas/")
r.json()
```

## Crear jornada

```python
# Crear jornada - caso exitoso
url = f"{BASE_URL}/jornadas/"
payload = {
    "fecha": "2026-06-03",
    "hora_inicio": "08:00:00",
    "hora_fin": "16:00:00",
    "empleado_id": 1,
    "cliente": "Cliente Ejemplo S.A.",
    "direccion": "Av. Corrientes 1234, CABA",
    "descripcion": "Jornada de limpieza general"
}
r = requests.post(url, json=payload)
print(r.status_code)
r.json()
```

```python
# Crear segunda jornada (sin descripcion)
url = f"{BASE_URL}/jornadas/"
payload = {
    "fecha": "2026-06-04",
    "hora_inicio": "09:00:00",
    "hora_fin": "17:30:00",
    "empleado_id": 3,
    "cliente": "Empresa del Norte",
    "direccion": "San Martín 500, Rosario",
    "descripcion": None
}
r = requests.post(url, json=payload)
print(r.status_code)
r.json()
```

```python
# Error: hora_fin <= hora_inicio
url = f"{BASE_URL}/jornadas/"
payload = {
    "fecha": "2026-06-05",
    "hora_inicio": "16:00:00",
    "hora_fin": "08:00:00",
    "empleado_id": 1,
    "cliente": "Cliente Test",
    "direccion": "Calle Falsa 123"
}
r = requests.post(url, json=payload)
print(r.status_code)
r.json()
```

```python
# Error: empleado inexistente
url = f"{BASE_URL}/jornadas/"
payload = {
    "fecha": "2026-06-05",
    "hora_inicio": "08:00:00",
    "hora_fin": "16:00:00",
    "empleado_id": 9999,
    "cliente": "Cliente Test",
    "direccion": "Calle Falsa 123"
}
r = requests.post(url, json=payload)
print(r.status_code)
r.json()
```

## Obtener jornada por ID

```python
r = requests.get(f"{BASE_URL}/jornadas/1")
print(r.status_code)
r.json()
```

```python
# Error: jornada inexistente
r = requests.get(f"{BASE_URL}/jornadas/9999")
print(r.status_code)
r.json()
```

## Editar jornada

```python
# Actualizar descripcion y hora_fin
url = f"{BASE_URL}/jornadas/1"
payload = {
    "hora_fin": "17:00:00",
    "descripcion": "Jornada extendida - limpieza profunda"
}
r = requests.patch(url, json=payload)
print(r.status_code)
r.json()
```

```python
# Reasignar a otro empleado
url = f"{BASE_URL}/jornadas/1"
payload = {
    "empleado_id": 3
}
r = requests.patch(url, json=payload)
print(r.status_code)
r.json()
```

## Filtrar jornadas

```python
# Filtrar por empleado
r = requests.get(f"{BASE_URL}/jornadas/filtrar", params={"empleado_id": 3})
r.json()
```

```python
# Filtrar por cliente (parcial, case-insensitive)
r = requests.get(f"{BASE_URL}/jornadas/filtrar", params={"cliente": "norte"})
r.json()
```

```python
# Filtrar por rango de fechas
r = requests.get(
    f"{BASE_URL}/jornadas/filtrar",
    params={"fecha_desde": "2026-06-01", "fecha_hasta": "2026-06-04"}
)
r.json()
```

```python
# Filtros combinados: empleado + rango de fechas
r = requests.get(
    f"{BASE_URL}/jornadas/filtrar",
    params={"empleado_id": 3, "fecha_desde": "2026-06-01", "fecha_hasta": "2026-06-30"}
)
r.json()
```

## Eliminar jornada

```python
r = requests.delete(f"{BASE_URL}/jornadas/2")
print(r.status_code)
```

```python
# Verificar que fue eliminada
r = requests.get(f"{BASE_URL}/jornadas/")
r.json()
```

## Validaciones de negocio (nuevas)

```python
import requests
# Setup: jornada de referencia para los tests siguientes → 201
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-10", "hora_inicio": "08:00:00", "hora_fin": "16:00:00",
    "empleado_id": 3, "cliente": "Cliente Ref", "direccion": "Av. Siempre Viva 742"
})
print(r.status_code)
ref_id = r.json()["id"]
r.json()
```

```python
# Error: empleado inactivo → 409  (ajustar empleado_id al ID inactivo real de la BD)
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-11", "hora_inicio": "08:00:00", "hora_fin": "16:00:00",
    "empleado_id": 2, "cliente": "Test Inactivo", "direccion": "Calle Falsa 123"
})
print(r.status_code)
r.json()
```

```python
# Error: duración supera 12 horas (06:00–20:00 = 14 h) → 422
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-11", "hora_inicio": "06:00:00", "hora_fin": "20:00:00",
    "empleado_id": 3, "cliente": "Test Duración", "direccion": "Calle X"
})
print(r.status_code)
r.json()
```

```python
# Error: jornada duplicada exacta (mismo empleado + fecha + horario) → 409
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-10", "hora_inicio": "08:00:00", "hora_fin": "16:00:00",
    "empleado_id": 3, "cliente": "Duplicada", "direccion": "Otra Dir"
})
print(r.status_code)
r.json()
```

```python
# Error: solapamiento parcial (ref es 08–16; esta empieza a las 10) → 409
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-10", "hora_inicio": "10:00:00", "hora_fin": "18:00:00",
    "empleado_id": 3, "cliente": "Test Solapa", "direccion": "Calle Overlap"
})
print(r.status_code)
r.json()
```

```python
# OK: jornada contigua (empieza justo donde termina la ref: 16:00) no solapa → 201
r = requests.post(f"{BASE_URL}/jornadas/", json={
    "fecha": "2026-06-10", "hora_inicio": "16:00:00", "hora_fin": "18:00:00",
    "empleado_id": 3, "cliente": "Cliente Contiguo", "direccion": "Av. Siempre Viva 742"
})
print(r.status_code)
contigua_id = r.json()["id"]
r.json()
```

```python
# Error update: hora_fin <= hora_inicio → 422
r = requests.patch(f"{BASE_URL}/jornadas/{ref_id}", json={"hora_inicio": "17:00:00", "hora_fin": "10:00:00"})
print(r.status_code)
r.json()
```

```python
# Error update: duración > 12 horas (05:00–20:00 = 15 h) → 422
r = requests.patch(f"{BASE_URL}/jornadas/{ref_id}", json={"hora_inicio": "05:00:00", "hora_fin": "20:00:00"})
print(r.status_code)
r.json()
```

```python
# Error update: solapamiento al extender hora_fin de ref (08–16) hasta 17:00,
# chocando con la jornada contigua (16–18) → 409
r = requests.patch(f"{BASE_URL}/jornadas/{ref_id}", json={"hora_fin": "17:00:00"})
print(r.status_code)
r.json()
```
