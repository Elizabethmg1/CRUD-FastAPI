```python
import requests
```

```python
r = requests.get("http://localhost:8000/empleados/")
r.json()
```

```python
#crear empleado
url = "http://localhost:8000/empleados"
payload = {"dni": "85895623", "nombre_completo": "Rogelio Johann", "observaciones": "Observaciones", "is_active": True}
r = requests.post(url, json=payload)
r.json()
```

```python
#modificar empleado
url = "http://localhost:8000/empleados/2"
payload = {"dni": "444555687", "nombre_completo": "Maria Gonzalez", "observaciones": "Observaciones", "is_active": True}
r = requests.patch(url, json=payload)
r.json()
```

```python
# eliminar empleado
url = "http://localhost:8000/empleados/1"
r = requests.delete(url)
r.status_code
```

```python
url = "http://localhost:8000/empleados/1/toggle-activo"
r = requests.patch(url)
print(r.status_code)
r.json()
```

```python
#busqueda empleados
url = "http://localhost:8000/empleados/buscar?q=85895623"
r = requests.get(url)
r.json()
```
